# Réseaux de neurones convolutifs modernes
:label:`chap_modern_cnn`

Maintenant que nous comprenons les bases de l'assemblage des CNN, faisons un tour d'horizon des architectures CNN modernes. Ce tour d'horizon est, par nécessité, incomplet, en raison de la pléthore de nouvelles conceptions passionnantes qui s'ajoutent sans cesse. Leur importance découle du fait qu'elles peuvent non seulement être utilisées directement pour des tâches de vision, mais qu'elles servent également de générateurs de caractéristiques de base pour des tâches plus avancées telles que le suivi :cite:`Zhang.Sun.Jiang.ea.2021`, la segmentation :cite:`Long.Shelhamer.Darrell.2015`, la détection d'objets :cite:`Redmon.Farhadi.2018` ou la transformation de style :cite:`Gatys.Ecker.Bethge.2016`. Dans ce chapitre, la plupart des sections correspondent à une architecture CNN significative qui a été à un moment donné (ou est actuellement) le modèle de base sur lequel de nombreux projets de recherche et systèmes déployés ont été construits. Chacun de ces réseaux a été brièvement une architecture dominante et beaucoup ont été vainqueurs ou finalistes de la [compétition ImageNet](https://www.image-net.org/challenges/LSVRC/) qui sert de baromètre des progrès de l'apprentissage supervisé en vision par ordinateur depuis 2010. Ce n'est que récemment que les Transformers ont commencé à supplanter les CNN, en commençant par :citet:`Dosovitskiy.Beyer.Kolesnikov.ea.2021` et suivis par le Swin Transformer :cite:`liu2021swin`. Nous aborderons ce développement plus tard dans le :numref:`chap_attention-and-transformers`. 

Bien que l'idée des réseaux de neurones *profonds* soit assez simple (empiler un ensemble de couches), les performances peuvent varier considérablement selon les architectures et les choix d'hyperparamètres. Les réseaux de neurones décrits dans ce chapitre sont le produit de l'intuition, de quelques aperçus mathématiques et de beaucoup de tâtonnements. Nous présentons ces modèles par ordre chronologique, en partie pour transmettre un sens de l'histoire afin que vous puissiez former vos propres intuitions sur la direction que prend le domaine et peut-être développer vos propres architectures. Par exemple, la normalisation par lots (batch normalization) et les connexions résiduelles décrites dans ce chapitre ont offert deux idées populaires pour l'entraînement et la conception de modèles profonds, qui ont toutes deux été depuis appliquées à des architectures au-delà de la vision par ordinateur.

Nous commençons notre tour d'horizon des CNN modernes avec AlexNet :cite:`Krizhevsky.Sutskever.Hinton.2012`, le premier réseau à grande échelle déployé pour battre les méthodes de vision par ordinateur conventionnelles sur un défi de vision à grande échelle ; le réseau VGG :cite:`Simonyan.Zisserman.2014`, qui utilise un certain nombre de blocs d'éléments répétitifs ; le network in network (NiN) qui convolue des réseaux de neurones entiers par patchs sur les entrées :cite:`Lin.Chen.Yan.2013` ; GoogLeNet qui utilise des réseaux avec des convolutions multi-branches :cite:`Szegedy.Liu.Jia.ea.2015` ; le réseau résiduel (ResNet) :cite:`He.Zhang.Ren.ea.2016`, qui reste l'une des architectures prêtes à l'emploi les plus populaires en vision par ordinateur ; les blocs ResNeXt :cite:`Xie.Girshick.Dollar.ea.2017` pour des connexions plus éparses ; et DenseNet :cite:`Huang.Liu.Van-Der-Maaten.ea.2017` pour une généralisation de l'architecture résiduelle. Au fil du temps, de nombreuses optimisations spéciales pour des réseaux efficaces ont été développées, telles que les décalages de coordonnées (ShiftNet) :cite:`wu2018shift`. Cela a abouti à la recherche automatique d'architectures efficaces telles que MobileNet v3 :cite:`Howard.Sandler.Chu.ea.2019`. Cela inclut également l'exploration de conception semi-automatique de :citet:`Radosavovic.Kosaraju.Girshick.ea.2020` qui a conduit au RegNetX/Y dont nous discuterons plus tard dans ce chapitre. Ce travail est instructif dans la mesure où il offre une voie pour marier le calcul par force brute avec l'ingéniosité d'un expérimentateur dans la recherche d'espaces de conception efficaces. On notera également le travail de :citet:`liu2022convnet` car il montre que les techniques d'entraînement (par exemple, les optimiseurs, l'augmentation de données et la régularisation) jouent un rôle pivot dans l'amélioration de la précision. Il montre également que des hypothèses de longue date, telles que la taille d'une fenêtre de convolution, peuvent nécessiter d'être revisitées, compte tenu de l'augmentation du calcul et des données. Nous aborderons ces questions et bien d'autres en temps voulu tout au long de ce chapitre.
